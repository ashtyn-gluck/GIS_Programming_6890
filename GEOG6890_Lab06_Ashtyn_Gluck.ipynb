## Exercise 1: Working with Text Files

Create a Python script that:

- Creates a text file named coordinates.txt
- Writes at least five latitude/longitude pairs to the file
- Reads the file back into Python
- Prints each coordinate pair on a separate line
- Uses a with open() statement

```python
#Create the coordinates
coordinates = [
    "32.6099, -85.4808", #Auburn
    "33.7490, -84.3880", #Atlanta
    "40.7128, -74.0050", #New York
    "34.0522, -118.2437", #Los Angeles
    "47.6062, -122.3321" #Seattle
]
```

```python
#Write coordinates to a text file
with open("coordinates.txt", "w") as file:
    for coordinate in coordinates:
        file.write(coordinate + "\n")

print("coordinates.txt has been created successfully.")
```

```python
#Read coordinates from the file
with open("coordinates.txt", "r") as file:
    for line in file:
        lat, lon = line.strip().split(",") #Remove the spaces and separate latitude and longitude

        print(f"Latitude: {lat}, Longitude: {lon}") #Print each coordinate pair
```

## Exercise 3: Handling File and Data Errors

Start with the following function:

Complete the function so that it:

- Opens a coordinate file
- Reads each line
- Converts latitude and longitude to float
- Handles FileNotFoundError
- Handles ValueError
- Continues processing if one line contains invalid data
- Prints "Processing finished." using finally


```python
def read_coordinates(filename):
    try: 
        with open(filename, "r") as file:
            for line in file:
                try:
                    lat, lon = line.strip().split(",")
    
                    lat = float(lat)
                    lon = float(lon)

                    print(f"Latitude: {lat}, Longitude: {lon}")

                except ValueError:
        
                    print(f"Invalid data: {line.strip()}")

    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found.")

    finally:
        print("Processing finished.")

read_coordinates("coordinates.txt")
```

## Exercise 4: Validate Geographic Coordinates
Create a function called:

`validate_coordinates(lat, lon)`

The function should return True only when:

`-90 ≤ latitude ≤ 90`

`-180 ≤ longitude ≤ 180`

Then combine the function with a file-processing script.

Your script should:

- Read coordinates from coordinates.txt
- Convert each value to float
- Check whether the coordinates are geographically valid
- Print valid coordinates
- Print a warning for invalid coordinates
- Handle formatting errors using exception handling


```python
#1. Create a function to validate coordinates
def validate_coordinates(lat, lon):
    if -90 <= lat <= 90 and -180 <= lon <= 180:
        return True
    else:
        return False
```

```python
#2. Read and process coordinates form the file
def process_coordinates(filename):

    try:
        with open(filename, "r") as file:
            for line in file:
                try:
                    lat, lon = line.strip().split(",")

                    lat = float(lat)
                    lon = float(lon)

                    if validate_coordinates(lat, lon):
                        print(
                            f"Latitude = {lat}, Longitude = {lon}"
                        )
                    else:
                        print(
                            f"Warning: Invalid geographic coordinates: "
                            f"{lat}, {lon}"
                        )
                except ValueError:
                    print(f"Warning: Invalid data: {line.strip()}")
    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found.")
        
    finally:
        print("Processing Finished.")
                    
process_coordinates("coordinates.txt")
```

## Exercise 5: Read Live GeoJSON Data from a URL
Use the requests library and this url = "https://earthquake.usgs.gov/earthquakes/feed/v1.0/summary/all_month.geojson"

Write a program that:

- Sends a request using requests.get()
- Uses raise_for_status() to check HTTP errors
- Converts the response using .json() into Python objects
- Extracts the "features" list
- Displays the first 10 earthquakes
- Prints:
    - Location
    - Magnitude
    - Latitude
    - Longitude
    - Depth

```python
import requests
```

```python
url = (
    "https://earthquake.usgs.gov/earthquakes/feed/v1.0/summary/all_month.geojson"
)

#1. Send a request to the URL
response = requests.get(url)

#2. Check for HTTP errors
response.raise_for_status()

#3. Convert the GeoJSON response to a python dictionary
earthquake_data = response.json()

#4. Extract the "features" list
features = earthquake_data["features"]

print(f"Total number of earthquakes: {len(features)}")
print()

#5. Display the first 10 earthquakes
for feature in features[:10]:

    #Get earthquake properties
    location = feature["properties"]["place"]
    magnitude = feature["properties"]["mag"]

    #GeoJSON coordinates are:
    longitude, latitude, depth = feature["geometry"]["coordinates"]

    #print earthquake information
    print(f"Location: {location}")
    print(f"Magnitude: {magnitude}")
    print(f"Latitude: {latitude}")
    print(f"Longitude: {longitude}")
    print(f"Depth: {depth} km")
    print("-" * 40)
    
```

```python
import csv

def validate_coordinates(latitude, longitude):
    if -90 <= latitude <= 90 and -180 <= longitude <= 180:
        return True
    else:
        return False

filename = "city_coordinates.csv"
total_records = 0
valid_records = 0
invalid_records = 0

try: #2. Use try and except to handle: FileNotFoundError, ValueError, Other unexpected errors
    with open("city_coordinates.csv", "r") as file, open("valid_cities.csv", "w", newline="") as outfile: #8. Write all valid records to:valid_cities.csv
        reader = csv.DictReader(file) #1. Use csv.DictReader() to read the input CSV file.

        fields = ["City", "Latitude", "Longitude", "Lat_Direction", "Lon_Direction"]
        writer = csv.DictWriter(outfile, fieldnames=fields)
        writer.writeheader()

        for row in reader:
            total_records += 1 #6. Count: Total records, Valid records, Invalid records
            try:
                city = row["City"]
                latitude = float(row["Latitude"]) #3. Convert latitude and longitude values to float.
                longitude = float(row["Longitude"])
                
                if validate_coordinates(latitude, longitude): #4. Validate coordinate ranges                  
                    if latitude >= 0: #7. Determine whether each valid city is in the Northern/Southern or Eastern/Western Hemisphere
                        lat_direction = "Northern Hemisphere"
                    else:
                        lat_direction = "Southern Hemisphere"
                    if longitude >= 0:
                        long_direction = "Eastern Hemisphere"
                    else:
                        long_direction = "Western Hemisphere"
                    print(f"{city}: Latitude = {latitude} ({lat_direction}), Longitude = {longitude} ({long_direction})")
                
                    writer.writerow({
                        "City": city,
                        "Latitude": latitude,
                        "Longitude": longitude,
                        "Lat_Direction": lat_direction,
                        "Lon_Direction": long_direction
                    })
                    valid_records += 1 #6. Count: Total records, Valid records, Invalid records
                else:
                    print(f"Invalid coordinates for {city}")
                    print("-"*40)
                    invalid_records += 1 #6. Count: Total records, Valid records, Invalid records
                    continue   #5. Skip invalid records without stopping the program.
                    
            except ValueError as e:
                print(f"Invalid data for {city}: {e}")
                invalid_records += 1
            
            print("-" *40)

except FileNotFoundError:
    print(f"Error: The file '{filename}' was not found.")
except Exception as e:
    print(f"Unexpected error: {e}")
finally:
    print("Processing finished.")
print("-" *40)
print(f"Number of valid records: {valid_records}")
print(f"Number of invalid records: {invalid_records}")
print(f"Number of total records: {total_records}")
```
