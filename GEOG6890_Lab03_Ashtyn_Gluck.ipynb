## Exercise 3: Manipulating Geographic Location Strings

- Create a string that represents the name of a geographic feature (e.g., `"Amazon River"`).
- Convert the string to lowercase and then to uppercase.
- Concatenate the string with the name of the country (e.g., `"Brazil"`) to create a full location name.
- Repeat the string three times, separating each repetition with a dash (`-`).

```python
site = "Amazon River"
```

```python
#Convert the string to lowercase and then to uppercase
lowercase_site = site.lower()
print(lowercase_site)
uppercase_site = site.upper()
print(uppercase_site)
```

```python
#Concatenate with a country name
country = "Brazil"
full_location = site + ", " + country
print(full_location)
```

```python
#Repeat three times with a dash
repeated_site = "-".join([site] * 3)
print(repeated_site)
```

## Exercise 4: Extracting and Formatting Coordinates

- Given a string with the format `"latitude, longitude"` (e.g., `"40.7128N, 74.0060W"`), extract the numeric values of latitude and longitude.
- Convert these values to floats and remove the directional indicators (`N`, `S`, `E`, `W`).
- Format the coordinates into a `POINT` WKT string (e.g., `"POINT(-74.0060 40.7128)"`).

```python
#coordinates
coordinates = "40.7128N, 74.0060W"
#split lat and long
latitude, longitude = coordinates.split(", ")
```

```python
#remove directional indicators
latitude_value = float(latitude[:-1]) #remove N
longitude_value = float(longitude[:-1]) #remove W
print(latitude_value)
print(longitude_value)
```

```python
#create WTK point
wkt = f"POINT({longitude_value}, {latitude_value})"
print(wkt)
```
