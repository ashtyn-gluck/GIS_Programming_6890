## Homework
Create a dictionary to store information for **three different cities** you have visited. For each city, include city name, coordinates, population. Calculate the centroid of these coordinates.

```python
#Create a dictionary to store information for three different cities I have visited with city name, coordinates, and population
cities = {
    "Columbus, OH": {
        "latitude": 39.9623,
        "longitude": -83.0007,
        "population": 983396
    },
    "Pensacola, FL": {
        "latitude": 30.4213,
        "longitude": -87.2169,
        "population": 53817
    },
    "Phoenix, AZ": {
        "latitude": 33.4484,
        "longitude": -112.0740,
        "population": 1665481
    }
}
print(cities)
```

```python
#Calculate the centroid of these coordinates
latitude_centroid = (cities["Columbus, OH"]["latitude"] + cities["Pensacola, FL"]["latitude"] + cities["Phoenix, AZ"]["latitude"]) / 3
longitude_centroid = (cities["Columbus, OH"]["longitude"] + cities["Pensacola, FL"]["longitude"] + cities["Phoenix, AZ"]["longitude"]) / 3
centroid = (latitude_centroid, longitude_centroid)

print("The centroid of the three cities is: ", centroid)
```
