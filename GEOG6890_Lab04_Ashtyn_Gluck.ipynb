## Exercise 1: Filter and Classify Geographic Coordinates
locations = [

    ("Auburn", 32.61, -85.48),
    
    ("Atlanta", 33.75, -84.39),
    
    ("Invalid_A", 95.00, -85.00),
    
    ("Birmingham", 33.52, -86.81),
    
    ("Invalid_B", 32.00, -190.00),
    
    ("Montgomery", 32.37, -86.30)
]

Tasks:
1. Loop through all locations.
2. Check whether each coordinate is valid:
     - Latitude: `-90` to `90`
     - Longitude:`-180` to `180`
3. If the coordinates are invalid, print a warning and use `continue`
4. For valid locations, determin whether they are:
   - North of 33°N
   - South of 33°N
   - Exactly at 33°N
5. Count the number of valid and invalid locations

```python
locations = [

("Auburn", 32.61, -85.48),

("Atlanta", 33.75, -84.39),

("Invalid_A", 95.00, -85.00),

("Birmingham", 33.52, -86.81),

("Invalid_B", 32.00, -190.00),

("Montgomery", 32.37, -86.30)
]
```

```python
invalid_locations = 0
valid_locations = 0

for city, lat, lon in locations: #Loop through all locations
    if lat <= -90 or lat >= 90 or lon <= -180 or lon >= 180: #Check whether each coordinate is valid
        print("Warning: coordinates invalid")
        invalid_locations += 1
        continue #If the coordinates are invalid, print a warning and use continue
    elif lat > 33: #For valid locations, determine whether they are north, south, or at 33 degrees north
        print(f"{city} is north of 33 degrees")
        valid_locations += 1
    elif lat == 33:
        print(f"{city} is exactly at 33 degrees")
        valid_locations += 1
    else:
        print(f"{city} is south of 33 degrees")
        valid_locations += 1

print("___________________________________")
print("The number of invalid locations: ", invalid_locations) #Count the number of valid or invalid locations
print("The number of valid locations: ", valid_locations)
```

## Exercise 3: Process elevation data with `continue` and `break`

elevations = [

    ("Auburn", 216),
    
    ("Montgomery", 73),
    
    ("Birmingham", 196),
    
    ("Unknown", None),
    
    ("Cheaha Mountain", 735),
    
    ("Atlanta", 320),
    
    ("Extreme_Point", 1500)
]

Tasks:

1. If elevation is `None`, print `"Missing data"` and use `continue`.
2. Classify valid elevations:
    - `< 100 m`: Low
    - `100–300 m`: Moderate
    - `301–700 m`: High
    - `> 700 m`: Very High
3. Sum the total of **valid** elevations.
4. Count valid locations.
5. If an elevation greater than 1000 m is encountered:
    - Print a warning.
    - Use `break` to stop processing.
6. Calculate the average elevation of all locations processed before the loop stopped.

```python
elevations = [

("Auburn", 216),

("Montgomery", 73),

("Birmingham", 196),

("Unknown", None),

("Cheaha Mountain", 735),

("Atlanta", 320),

("Extreme_Point", 1500)
]
```

```python
valid_elevations = 0
total_elevation = 0

for city, elevation in elevations: #If elevation is None, print "Missing data" and use continue.
    if elevation is None:
        print("Missing Data")
        continue
        
    if elevation > 1000: #If an elevation greater than 1000 m is encountered:Print a warning.Use break to stop processing.
        print("Warning: Greater than 1000m")
        break
        
    if elevation < 100: #Classify the valid elevations
        print(f"{city}:", "Low Elevation")
    elif elevation <= 300:
        print(f"{city}:","Moderate Elevation")
    elif elevation <= 700:
        print(f"{city}:","High Elevation")
    else:
        print(f"{city}:","Very High Elevation")
    valid_elevations += 1 #Count the total of valid elevations
    total_elevation += elevation #Sum the total of valid elevations
        
if valid_elevations > 0: #Calculate the average elevation before the loop is stopped
    average_elevation = total_elevation / valid_elevations

print("___________________________________")
print("Total number of valid elevations: ", valid_elevations)
print("Average elevation: ", f"{average_elevation:.2f}")
```

## Exercise 4: Search a Coordinate Dataset with a `while` Loop

cities = ["Auburn", "Montgomery", "Birmingham", "Atlanta", "Mobile"]

latitudes = [32.61, 32.37, 33.52, 33.75, 30.69]

longitudes = [-85.48, -86.30, -86.81, -84.39, -88.04]


target = "Atlanta"

Tasks:

1. Use a `while` loop. DO NOT use a `for` loop.
2. Search for target, print each city as it is checked.
3. When the target is found:
    - Print its latitude and longitude.
    - Use `break`.
4. Count how many locations were searched.


```python
cities = ["Auburn", "Montgomery", "Birmingham", "Atlanta", "Mobile"]

latitudes = [32.61, 32.37, 33.52, 33.75, 30.69]

longitudes = [-85.48, -86.30, -86.81, -84.39, -88.04]

target = "Atlanta"
```

```python
i = 0
searched_locations = 0

while i < len(cities): #Use a while loop
    city = cities[i]
    searched_locations += 1 #Count how many locations were searched

    print("Checking:" f"{city}") #Search for target, print each city as it is checked

    if city == target: #When the target is found Print latitude and longitude and use break
        print(f"{city}, Latitude: {latitudes[i]}, Longitude: {longitudes[i]}")
        break
    i += 1 

print("_________________")
print("Locations Searched: ", searched_locations)
```
